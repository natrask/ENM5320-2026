This jupyter notebook will show how to construct a simple cross-attention transformer that can be used to construct conditional neural fields.
$$y = F(x|z)$$
Here, $x,y$ are input/output labels and $z$ denotes a conditioning variable which modulates the input/output relationship.

This code can be adopted into your final projects - in general this is drastically easier to train than a standard MLP, in addition to providing support for conditioning (if you want/need that).

Some key aspects:

*   The code consists of a `CrossAttnDecoderLayer` block which can be stacked into $N_{blocks}$ blocks to construct deeper architectures.
*   Transformers are in general very data-hungry, and in scientific settings we will rarely have enough data (consider e.g. GPT trained on Wikipedia, vs the 10 experiments you might be able to do in the lab). To mitigate this, dropout is crucial to regularize and prevent overfitting. The `update_dropout_rate` function can be used to control this. In practice you may want to train with dropout turned on, and then implement a scheduler to ease it off as training progresses.
*   In addition to the architecture, there have been a number of works in the past year (2025) where people have explored approximate second-order optimizer (i.e. automatic differentiation is used to construct approximate Hessians, rather that working with a momentum-adjusted stepsize in gradient descent). This code uses Shampoo, which we use in my group. This requires installing Pytorch's torch-optimizer package which includes "non-standard" optimizers that haven't yet made it into the main torch branch.

The code below shows how to instantiate the model, initialize the shampoo optimizer, and solve a simple regression task. No effort has been made to optimize this (e.g. layers, learning rates, etc) so you may need to fiddle with some parameters to get best performance on your problem.

## 2. Architecture: Cross-attention decoder

We define two classes:

1. **`CrossAttnDecoderLayer`** — A single transformer decoder block. Each block performs:
   - **Cross-attention**: the evaluation points $x$ (query) attend to the conditioning variable $z$ (key/value). Because $z$ is a single "global" token, this is akin to broadcasting the conditioning information to every point in the sequence.
   - **Feed-forward network (FFN)**: a two-layer MLP with GELU activation that post-processes the attended features.
   - Both sub-layers use **pre-norm residual connections** (LayerNorm applied *before* each sub-layer, then added back via a skip connection).

2. **`CrossAttnTransformer`** — Stacks $N_\text{blocks}$ decoder layers. The full forward pass is:
   $$x \;\xrightarrow{\text{Linear}}\; \mathbf{h}^{(0)} \;\xrightarrow{\text{Block}_1}\; \mathbf{h}^{(1)} \;\cdots\; \xrightarrow{\text{Block}_N}\; \mathbf{h}^{(N)} \;\xrightarrow{\text{Linear}}\; \hat{y}$$
   where $z$ is independently embedded and injected at every block via cross-attention.

**Key design choice:** because queries come from $x$ and keys/values come from $z$, this is *cross*-attention (not self-attention). The model learns to modulate its output field $\hat y(x)$ based on the conditioning variable $z$.

```python
class CrossAttnDecoderLayer(nn.Module):
    def __init__(self, d_model, num_heads, dropout_rate, activation, use_norms):
        super().__init__()

        self.d_model = d_model
        self.num_heads = num_heads
        self.dropout_rate = dropout_rate
        self.activation = activation
        self.use_norms = use_norms

        self.multihead_attn = nn.MultiheadAttention(
            embed_dim=d_model,
            num_heads=num_heads,
            dropout=dropout_rate,
            batch_first=True,
        )

        self.ffn = nn.Sequential(
            nn.Linear(d_model, d_model * 2),
            activation,
            nn.Dropout(dropout_rate),
            nn.Linear(d_model * 2, d_model),
        )

        self.dropout = nn.Dropout(dropout_rate)
        self.norm1 = nn.LayerNorm(d_model)
        self.norm2 = nn.LayerNorm(d_model)

    def update_dropout_rate(self, new_rate):
        self.dropout_rate = new_rate

        self.multihead_attn.dropout = new_rate

        for module in self.ffn:
            if isinstance(module, nn.Dropout):
                module.p = new_rate

        self.dropout.p = new_rate

    def forward(self, x, z_embedding):
        # input shapes:
        # x: [batch_size (optional), seq_len (number of x eval points), d_model]
        # z_embedding: [batch_size (optional), d_model]
        # in order to make z_embedding work as a "global" token, we need to add a dummy seq_len dimension of 1
        z_embedding = z_embedding.unsqueeze(-2)  # [batch_size (optional), 1, d_model]
        residual = x
        if self.use_norms:
            x_norm = self.norm1(x)
        else:
            x_norm = x

        attn_output, _ = self.multihead_attn(
            query=x_norm,
            key=z_embedding,
            value=z_embedding,
            need_weights=False,
        )

        x = residual + self.dropout(attn_output)

        residual = x
        if self.use_norms:
            x_norm = self.norm2(x)
        else:
            x_norm = x
        x = residual + self.dropout(self.ffn(x_norm))

        return x


class CrossAttnTransformer(nn.Module):
    def __init__(
        self,
        problem_dim,
        output_dim,
        z_dim,
        num_blocks=3,
        latent_dim=128,
        num_heads=4,
        dropout_rate=0.1,
        activation=nn.GELU(),
        use_norms=True,
    ):
        super().__init__()

        self.problem_dim = problem_dim
        self.output_dim = output_dim
        self.z_dim = z_dim
        self.num_blocks = num_blocks
        self.latent_dim = latent_dim
        self.num_heads = num_heads
        self.dropout_rate = dropout_rate
        self.activation = activation
        self.use_norms = use_norms

        self.x_embedding = nn.Linear(self.problem_dim, self.latent_dim)
        self.z_embedding = nn.Linear(self.z_dim, self.latent_dim)

        self.transformer_blocks = nn.ModuleList(
            [
                CrossAttnDecoderLayer(
                    d_model=self.latent_dim,
                    num_heads=num_heads,
                    dropout_rate=dropout_rate,
                    activation=activation,
                    use_norms=use_norms,
                )
                for _ in range(num_blocks)
            ]
        )

        self.final_projection = nn.Linear(self.latent_dim, self.output_dim)

    def update_dropout_rate(self, new_rate):
        self.dropout_rate = new_rate
        for block in self.transformer_blocks:
            block.update_dropout_rate(new_rate)

    def forward(self, x, z):
        x_emb = self.x_embedding(x)
        z_emb = self.z_embedding(z)

        for block in self.transformer_blocks:
            x_emb = block(x_emb, z_emb)

        return self.final_projection(x_emb)
```

## 3. Instantiate the model

We create a `CrossAttnTransformer` with:
- `problem_dim=1`, `output_dim=1`, `z_dim=1` — scalar input/output/conditioning
- `num_blocks=3` — three stacked decoder layers
- `latent_dim=128` — hidden dimension
- `num_heads=4` — multi-head attention splits the 128-dim space into four 32-dim heads
- `dropout_rate=0` — **dropout is turned off** for this simple demo (see the intro for guidance on using dropout for larger problems)

**Try it yourself:** change `num_blocks` or `latent_dim` and observe the effect on convergence speed and final accuracy.

```python
model = CrossAttnTransformer(
    problem_dim=1,
    output_dim=1,
    z_dim=1,
    num_blocks=3,
    latent_dim=128,
    num_heads=4,
    dropout_rate=0,
    activation=nn.GELU(),
    use_norms=True,
).to(device)
```

## 4. Generate training and validation data

To illustrate standard regression, we learn the family of functions
$$y(x \mid z) = \sin(z\, x), \qquad x \in [-\pi,\pi],\quad z \in [1,5].$$

Here $z$ plays the role of a **conditioning variable** (wavenumber/angular frequency). We sample 12 evenly-spaced $z$ values for **training** and use the 11 midpoints between them for **validation** — this tests whether the model can *interpolate* in $z$-space, not just memorize the training frequencies.

```python
# z is going to be the angular frequency from 1 to 5, say
z = torch.linspace(1, 5, 12).unsqueeze(-1)

# x are the evaluation points
x = torch.linspace(-torch.pi, torch.pi, 256)

# y are the target functions, sin(z * x), evaluated at the x points
y = torch.stack([torch.sin(z[i] * x) for i in range(z.shape[0])]).unsqueeze(-1)

# we also build a validation set in between the z values:
z_val = (z[1:] + z[:-1]) / 2.0
y_val = torch.stack([torch.sin(z_val[i] * x) for i in range(z_val.shape[0])]).unsqueeze(
    -1
)
```

Move everything to the compute device and wrap in `DataLoader`s for batched iteration. Note that `batch_size=12` equals the full training set — this is fine for such a small dataset.

```python
print(f"Training:   z.shape={z.shape}, y.shape={y.shape}")
print(f"Validation: z_val.shape={z_val.shape}, y_val.shape={y_val.shape}")

y, z = y.to(device), z.to(device)
train_dataset = torch.utils.data.TensorDataset(z, y)
train_loader = DataLoader(train_dataset, batch_size=12, shuffle=True)

z_val, y_val = z_val.to(device), y_val.to(device)
val_dataset = torch.utils.data.TensorDataset(z_val, y_val)
val_loader = DataLoader(val_dataset, batch_size=len(z_val), shuffle=False)

x = x.to(device)
```

## 5. Optimizer: Shampoo vs Adam

[Shampoo](https://arxiv.org/abs/1802.09568) is an approximate second-order optimizer: it maintains Kronecker-factored preconditioners that approximate the inverse Hessian, giving it better curvature information than first-order methods like Adam. This can lead to faster convergence, especially for transformer architectures.

A commented-out Adam configuration is included below for comparison. **Exercise:** uncomment the Adam block (and comment out Shampoo), re-run training, and compare:

```python
optimizer = optim.Shampoo(
    [
        {
            "params": model.parameters(),
            "lr": 0.01,
            "weight_decay": 0,
            "momentum": 0,
            "update_freq": 1,
        }
    ]
)
# optimizer = torch.optim.Adam(
#     [
#         {
#             "params": model.parameters(),
#             "lr": 1e-5,
#             "weight_decay": 0,
#             "momentum": 0,
#         }
#     ]
# )
```

## 6. Checkpointing utilities

Save and load model/optimizer state so you can resume training without starting from scratch. This is good practice for any training run that takes more than a few minutes.

```python
def save(train_loss_history, val_loss_history):
    with open("cross_attn_transformer_test.pkl", "wb") as f:
        pickle.dump(
            {
                "model": model.state_dict(),
                "optimizer": optimizer.state_dict(),
                "train_loss_history": train_loss_history,
                "val_loss_history": val_loss_history,
            },
            f,
        )


def read():
    with open("cross_attn_transformer_test.pkl", "rb") as f:
        data = pickle.load(f)
        model.load_state_dict(data["model"])
        optimizer.load_state_dict(data["optimizer"])
        train_loss_history = data["train_loss_history"]
        val_loss_history = data["val_loss_history"]
    return train_loss_history, val_loss_history
```

```python
# read in checkpoint if it exists
try:
    with open("cross_attn_transformer_test.pkl", "rb") as f:
        data = pickle.load(f)
        model.load_state_dict(data["model"])
        optimizer.load_state_dict(data["optimizer"])
        train_loss_history = data["train_loss_history"]
        val_loss_history = data["val_loss_history"]
except FileNotFoundError:
    train_loss_history = []
    val_loss_history = []
    print("No checkpoint found, starting from scratch.")
```

## 7. Training loop

The training loop below runs for 1000 epochs. A few things to observe:

- **`x_expanded`**: the shared evaluation grid `x` is broadcast to match the batch dimension of `z`. This is how every sample in the batch is evaluated at the *same* set of input points — a natural structure for neural fields.
- **Optimizer timing**: we explicitly measure how much wall-clock time is spent in `optimizer.step()`. For second-order methods like Shampoo, this fraction can be significant — the per-step cost is higher but you may need fewer steps overall.
- **Validation**: performed inside `torch.no_grad()` to save memory and computation. The commented-out `model.eval()` line is left intentionally — since `dropout_rate=0`, train/eval modes are equivalent here. For nonzero dropout you would want to uncomment it.

> **Tip:** if 1000 epochs is too slow on CPU, reduce to ~200 for a quick sanity check.

```python
time_in_optimizer = 0.0

overall_time_start = time.time()
for epoch in range(0, 1000):
    model.train()
    batch_loss = 0.0
    for i, (z, y) in enumerate(train_loader):
        optimizer.zero_grad()

        # Forward pass
        x_expanded = x.unsqueeze(0).unsqueeze(-1).expand(z.shape[0], -1, -1)

        y_pred = model(x_expanded, z)

        # Compute loss
        loss = nn.functional.mse_loss(y_pred, y)

        batch_loss += loss.item()

        # Backward pass and optimization
        loss.backward()
        time_start = time.time()
        optimizer.step()
        time_end = time.time()

        time_in_optimizer += time_end - time_start

    batch_loss /= len(train_loader)
    train_loss_history.append(batch_loss)

    # Validation step
    # model.eval()
    with torch.no_grad():
        batch_loss = 0.0
        for z_val, y_val in val_loader:
            x_expanded = x.unsqueeze(0).unsqueeze(-1).expand(z_val.shape[0], -1, -1)

            y_val_pred = model(x_expanded, z_val)

            batch_loss += nn.functional.mse_loss(y_val_pred, y_val).item()
        batch_loss /= len(val_loader)

    val_loss_history.append(batch_loss)

    print(
        f"Epoch {epoch}: Train Loss: {train_loss_history[-1]:.4e}, Val Loss: {val_loss_history[-1]:.4e}"
    )

    if epoch % 1000 == 0:
        save(train_loss_history, val_loss_history)
        print("Checkpoint saved.")

overall_time_end = time.time()

frac_in_optimizer = time_in_optimizer / (overall_time_end - overall_time_start)
print(f"Fraction of time in optimizer: {frac_in_optimizer:.6f}")
```

## 8. Results

### Loss curves
Plot training and validation loss on a log-log scale. Both curves should decrease together — if validation loss plateaus or increases while training loss keeps dropping, that signals overfitting (try increasing `dropout_rate`).

```python
plt.figure(figsize=(8, 5))
plt.plot(train_loss_history, label="Train Loss")
plt.plot(val_loss_history, label="Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("MSE Loss")
plt.title("Training and Validation Loss (log-log scale)")
plt.legend()
plt.loglog()
plt.grid(True, alpha=0.3)
plt.show()
```

### Training set predictions
Compare the model's output (predicted) against the ground truth (true) for each training $z$ value. The model should match these well since it trained on them.

```python
with torch.no_grad():
    z, y = train_dataset.tensors
    x_expanded = x.unsqueeze(0).unsqueeze(-1).expand(z.shape[0], -1, -1)
    y_pred = model(x_expanded, z)

    fig, axs = plt.subplots(3, 4, figsize=(15, 10))
    fig.suptitle("Training Set: True vs Predicted", fontsize=14)
    for i in range(12):
        ax = axs[i // 4, i % 4]
        ax.plot(x.cpu(), y[i].cpu(), label="True")
        ax.plot(x.cpu(), y_pred[i].cpu(), "--", label="Predicted")
        ax.set_title(f"z = {z[i].item():.2f}")
        if i == 0:
            ax.legend(fontsize=8)
    plt.tight_layout()
    plt.show()
```

### Validation set predictions (interpolation in $z$)
These are the 11 midpoint $z$ values the model *never* saw during training. Good performance here demonstrates that the model has learned a smooth, generalizable mapping in $z$-space rather than memorizing individual frequencies.

```python
with torch.no_grad():
    for z_val, y_val in val_loader:
        x_expanded = x.unsqueeze(0).unsqueeze(-1).expand(z_val.shape[0], -1, -1)
        y_val_pred = model(x_expanded, z_val)

        fig, axs = plt.subplots(3, 4, figsize=(15, 10))
        fig.suptitle("Validation Set (Interpolated z): True vs Predicted", fontsize=14)
        for i in range(z_val.shape[0]):
            ax = axs[i // 4, i % 4]
            ax.plot(x.cpu(), y_val[i].cpu(), label="True")
            ax.plot(x.cpu(), y_val_pred[i].cpu(), "--", label="Predicted")
            ax.set_title(f"z = {z_val[i].item():.2f}")
            if i == 0:
                ax.legend(fontsize=8)
        # hide the unused subplot (11 values in a 3x4 grid)
        axs[2, 3].set_visible(False)
        plt.tight_layout()
        plt.show()
```
